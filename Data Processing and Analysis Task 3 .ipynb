**_1: Employee Performance Bonus Eligibility_**

```python
# Dictionary holding employee performance ratings
performance_data = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Determining the maximum score
best_score = max(performance_data.values())

# Collecting employees who achieved the top score
eligible_employees = []

for emp_name, emp_score in performance_data.items():
    if emp_score == best_score:
        eligible_employees.append(emp_name)

# Displaying the final output
print(
    "Top Performers Eligible for Bonus:",
    ", ".join(eligible_employees),
    "(Score:", best_score, ")"
)
```

**_2: Search Query Keyword Analysis_**

```python
# Search query text
search_text = "Buy mobile phone buy phone online"

# Converting text to lowercase for uniformity
search_text = search_text.lower()

# Splitting text into individual terms
terms = search_text.split()

# Dictionary to store term frequencies
term_count = {}

for t in terms:
    term_count[t] = term_count.get(t, 0) + 1

# Extracting terms that appear more than once
common_terms = {}
for t, c in term_count.items():
    if c > 1:
        common_terms[t] = c

# Displaying repeated keywords
print(common_terms)
```

**_3: Sensor Data Validation_**

```python
# Sensor values collected over time
readings = [3, 4, 7, 8, 10, 12, 5]

even_readings = []

# Identifying even sensor values with their index
for index, reading in enumerate(readings):
    if reading % 2 == 0:
        even_readings.append((index, reading))

# Displaying valid sensor readings
print("Valid Sensor Readings (Hour, Value):")
print(even_readings)
```

**_ 4: Email Domain Usage Analysis_**

```python
# List of user email addresses
email_list = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

domain_usage = {}

# Extracting and counting email domains
for mail in email_list:
    mail_domain = mail.split("@")[-1]
    domain_usage[mail_domain] = domain_usage.get(mail_domain, 0) + 1

user_count = len(email_list)

# Displaying domain-wise usage percentage
for d, c in domain_usage.items():
    usage_percent = (c / user_count) * 100
    print(f"{d}: {int(usage_percent)}%")
```

**_5: Sales Spike Detection_**

```python
# Daily sales figures
daily_sales = [1200, 1500, 900, 2200, 1400, 3000]

# Computing the mean sales value
avg_value = sum(daily_sales) / len(daily_sales)

# Setting spike limit (30% above average)
spike_limit = avg_value * 1.3

# Identifying days with sales spikes
for idx, amount in enumerate(daily_sales, start=1):
    if amount > spike_limit:
        print(f"Day {idx}: {amount}")
```

**_ 6: Duplicate User ID Detection_**

```python
# Collection of user identifiers
users = ["user1", "user2", "user1", "user3", "user1", "user3"]

user_frequency = {}

# Tracking how many times each ID appears
for uid in users:
    user_frequency[uid] = user_frequency.get(uid, 0) + 1

# Printing only IDs that appear more than once
for uid, freq in user_frequency.items():
    if freq > 1:
        print(f"{uid} → {freq} times")
```
