```python
#1
#Predefined login details
saved_user = "admin"
saved_pass = "1234"

#Taking input from user
input_user = input("Enter username: ")
input_pass = input("Enter password: ")

#Verifying credentials
if input_user == saved_user:
    if input_pass == saved_pass:
        print("Login Successful")
    else:
        print("Invalid Credentials")
else:
    print("Invalid Credentials")
```

```python
#2
#List of student scores
scores = [45, 78, 90, 33, 60]

passed = 0
failed = 0

#Checking each score
for score in scores:
    if score >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

#Displaying summary
total_students = len(scores)
print("Total Students:", total_students)
print("Pass Students:", passed)
print("Fail Students:", failed)
```

```python
#3
#List containing names with extra spaces and mixed cases
name_list = [" Alice ", "bob", " CHARLIE "]

#Cleaning spaces from each name
trimmed_names = []
for n in name_list:
    trimmed_names.append(n.strip())

print("After removing spaces:", trimmed_names)

#Converting cleaned names to lowercase
final_list = []
for n in trimmed_names:
    final_list.append(n.lower())

print("After converting to lowercase:", final_list)

#Displaying final cleaned names
print("Cleaned Names List:", final_list)

```

```python
#4
#List containing user texts
text_list = ["Hi", "Welcome to the platform", "OK"]

#Processing each text
for text in text_list:
    length_of_text = len(text)

    print("Message:", text)
    print("Length:", length_of_text)

    #Checking message size
    if length_of_text > 10:
        print("Status: Long Message")
    else:
        print("Status: Short Message")

    print("-" * 22)
```

```python
#5
#System log records
log_entries = ["INFO", "ERROR", "WARNING", "ERROR"]

errors_found = 0

#Counting error messages
for entry in log_entries:
    if entry == "ERROR":
        errors_found = errors_found + 1

#Displaying result
print("Total ERROR entries:", errors_found)
```
