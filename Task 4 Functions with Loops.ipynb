**_1: Mobile Recharge Validation System_**

```python
# Function to check recharge validity
def check_recharge(value):
    plans_available = [199, 299, 399, 599]

    if value < 50:
        return False, "Recharge amount should be ₹50 or more."
    if value not in plans_available:
        return False, "Invalid plan selected. Choose from 199, 299, 399, 599."
    return True, "Recharge Successful!"


# Loop until a valid recharge is done
while True:
    user_amount = int(input("Enter recharge amount: ₹"))

    status, response = check_recharge(user_amount)
    print(response)

    if status:
        break
    print("Please try again.\n")
```

**_ 2: Inventory Reorder Alert System_**

```python
# Function to verify stock availability
def verify_stock(items):
    for item_name, count in items.items():
        if count < 15:
            print(f"{item_name}: Reorder Alert")
        else:
            print(f"{item_name}: Stock OK")


# Current inventory details
store_inventory = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 12,
    "Monitor": 30
}

# Running inventory check
verify_stock(store_inventory)
```

**_3: Student Result Processing System_**

```python
# Function to evaluate student performance
def evaluate_student(scores):
    total_score = 0

    # Summing all marks
    for s in scores:
        total_score += s

    # Finding average marks
    avg_score = total_score / len(scores)

    # Checking pass or fail status
    if avg_score >= 50:
        return "Pass"
    return "Fail"


# Marks obtained by the student
marks_list = [45, 60, 70, 55]

# Displaying final result
final_result = evaluate_student(marks_list)
print("Result:", final_result)
```

**_4: Cab Fare Estimator with Retry Option_**

```python
# Function to compute taxi fare
def compute_fare(km_travelled, peak_time):
    start_charge = 50
    rate_per_km = 12

    total_amount = start_charge + (km_travelled * rate_per_km)

    if peak_time:
        total_amount += total_amount * 0.25

    return total_amount


# Loop to allow multiple calculations
while True:
    kms = float(input("Enter distance in km: "))

    peak_choice = input("Is it peak hour? (yes/no): ").strip().lower()
    peak_flag = (peak_choice == "yes")

    fare_amount = compute_fare(kms, peak_flag)
    print(f"Total Fare: ₹{fare_amount:.2f}")

    again = input("Do you want to calculate again? (yes/no): ").strip().lower()
    if again != "yes":
        break
```

**_5: Employee Attendance Eligibility Checker _**

```python
# Function to check attendance eligibility
def check_attendance(attendance):
    present_days = 0

    # Count present days using loop
    for day in attendance:
        if day == "P":
            present_days += 1

    # Calculate attendance percentage
    attendance_percentage = (present_days / len(attendance)) * 100

    # Determine eligibility
    if attendance_percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


# Attendance data
employee_attendance = ["P", "P", "A", "P", "P", "A", "P"]

# Check and display result
result = check_attendance(employee_attendance)
print("Attendance Status:", result)
```

**_6: Password Strength Checker_**

```python
# Function to evaluate password strength
def evaluate_password(pwd):
    contains_number = False
    contains_special = False
    allowed_specials = "@#$"

    # Check minimum length requirement
    if len(pwd) < 8:
        return "Weak Password"

    # Scanning each character
    for ch in pwd:
        if ch.isdigit():
            contains_number = True
        if ch in allowed_specials:
            contains_special = True

    # Final decision
    if contains_number and contains_special:
        return "Strong Password"
    return "Weak Password"


# Sample password check
test_password = "Pass@123"
strength = evaluate_password(test_password)
print("Password Strength:", strength)
```
