**_Problem 1: Unique words in a Sentence_**

```python
# Taking sentence input from user
sentence = input("Enter sentence: ")

# Step 1: Split sentence into words
words = sentence.split()

# Step 2: Convert list to set to get unique words
unique_words = set(words)

# Step 3: Display result
print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)
```

**_Problem 2: Highest Salary from Employee Data_**

```python
# Dictionary storing employee salaries
staff_salaries = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

top_salary = 0
top_employee = None

# Identifying the highest paid employee
for emp, pay in staff_salaries.items():
    if pay > top_salary:
        top_salary = pay
        top_employee = emp

# Printing the result
print("Highest Salary:", top_employee, "-", top_salary)
```

**_Problem 3: Find Maximum and Minimum Values_**

```python
# Collection of numeric values
values = [45, 22, 89, 10, 66]

# Setting initial highest and lowest values
highest = values[0]
lowest = values[0]

# Evaluating each number
for val in values:
    if val > highest:
        highest = val
    elif val < lowest:
        lowest = val

# Showing the results
print("List:", values)
print("Max:", highest)
print("Min:", lowest)
```

**_Problem 4: Count Products Above a Price Threshold_**

```python
# Prices of available products
product_prices = [450, 1200, 899, 1500, 300]

above_limit = 0

# Checking prices greater than 1000
for amount in product_prices:
    if amount > 1000:
        above_limit = above_limit + 1

# Showing the final count
print("Products above 1000:", above_limit)
```

**_Problem 5: Calculate Attendance Percentage_**

```python
# Record of attendance status
attendance_record = ["P", "P", "A", "P", "P"]

present_days = 0

# Calculating number of days present
for status in attendance_record:
    if status == "P":
        present_days = present_days + 1

# Computing attendance percentage
days_count = len(attendance_record)
percentage = (present_days / days_count) * 100

# Displaying the result
print("Attendance Percentage:", percentage)
```

**_Problem 7: Count Character Frequency_**

```python
# Input word
word = "python"

# Dictionary to track letter counts
frequency = {}

# Counting each character in the word
for letter in word:
    if letter in frequency:
        frequency[letter] = frequency[letter + 0] + 1
    else:
        frequency[letter] = 1

# Printing the frequency dictionary
print(frequency)
```

**_Problem 9: Check if a Key Exists in a Dictionary_**

```python
# Employee information stored in a dictionary
employee_details = {
    "Employee": "Ravi",
    "Salary": 50000,
    "Department": "IT"
}

# Verifying presence of a key
if "Employee" in employee_details.keys():
    print("Employee exists")
else:
    print("Employee does not exist")
```

**_Problem 10: Calculate Average Marks_**

```python
# List of marks
marks = [70, 75, 80, 65]

total_marks = 0

# Calculate total marks
for m in marks:
    total_marks += m

# Calculate average
average_marks = total_marks / len(marks)

# Display result
print("Average Marks:", average_marks)
```
